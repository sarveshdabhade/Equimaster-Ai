```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os

# 1. Load one file to inspect
ticker = "RELIANCE.NS"
file_path = f"../data/{ticker}.csv"
if not os.path.exists(file_path):
    file_path = f"data/{ticker}.csv"  # fallback when run from project root (Cursor)

if os.path.exists(file_path):
    print(f"Found data for {ticker}")
    # yfinance CSV has 2 metadata rows; then: Date, Close, High, Low, Open, Volume
    df = pd.read_csv(file_path, skiprows=2, names=["Date", "Close", "High", "Low", "Open", "Volume"],
                     parse_dates=["Date"], index_col=0)
    df["Close"] = pd.to_numeric(df["Close"], errors="coerce")
    df = df.dropna(subset=["Close"])
else:
    print("File not found! Check your data_loader.py")
```

```python
# 2. Look at the raw data
print("Data Shape:", df.shape)  # Rows, Columns
print("\nFirst 5 Rows:")
display(df.head())

print("\nMissing Values Check:")
print(df.isnull().sum())  # This must be 0 for all columns!
print("\nZeros in Close:", (df["Close"] == 0).sum())
```

```python
# 3. Visual Sanity Check
plt.figure(figsize=(12, 6))
plt.plot(df['Close'], label='Close Price', color='blue', linewidth=1)
plt.title(f"{ticker} - 10 Year Price History")
plt.xlabel("Days")
plt.ylabel("Price (INR)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
